Neste notebook, vamos estender nosso estudo de transformações de Lorentz em $1+1$ dimensões para $3+1$ dimensões. O objetivo principal é investigar o que acontece quando compomos dois **boosts em direções perpendiculares** (por exemplo, eixo $x$ e eixo $y$).

Verificaremos que a composição de dois boosts não-colineares gera não apenas um novo boost, mas também uma **rotação espacial pura** — um fenômeno puramente relativístico conhecido como **Rotação de Wigner** (relacionado à precessão de Thomas).

## 1. Configurando a métrica 4D e as funções de boost

```python
import numpy as np

# Métrica de Minkowski com assinatura (-, +, +, +)
eta = np.diag([-1.0, 1.0, 1.0, 1.0])

def boost_x(v):
    gamma = 1.0 / np.sqrt(1.0 - v**2)
    L = np.eye(4)
    L[0, 0] = gamma
    L[0, 1] = -gamma * v
    L[1, 0] = -gamma * v
    L[1, 1] = gamma
    return L

def boost_y(v):
    gamma = 1.0 / np.sqrt(1.0 - v**2)
    L = np.eye(4)
    L[0, 0] = gamma
    L[0, 2] = -gamma * v
    L[2, 0] = -gamma * v
    L[2, 2] = gamma
    return L
```

## 2. Compondo os boosts e testando o intervalo

```python
v1 = 0.6
v2 = 0.6

# Composição: Lx seguido de Ly
Lambda = boost_y(v2) @ boost_x(v1)

# O intervalo deve se manter invariante: L^T eta L = eta
residuo = Lambda.T @ eta @ Lambda - eta
print(f"Resíduo da invariância do intervalo: {np.abs(residuo).max():.2e}")
```

## 3. Checando a rotação e calculando o ângulo

```python
# Pegando o bloco espacial 3x3
R_spatial = Lambda[1:, 1:]

# Se fosse um boost puro, a matriz espacial seria simétrica. Vamos testar:
print("É simétrica?", np.allclose(R_spatial, R_spatial.T))

# Fórmula analítica para v1 = v2 = 0.6
gamma1 = 1.0 / np.sqrt(1.0 - v1**2)
gamma2 = 1.0 / np.sqrt(1.0 - v2**2)
tan_theta = (gamma1 * gamma2 * v1 * v2) / (gamma1 + gamma2)
theta_deg = np.degrees(np.arctan(tan_theta))

print(f"Ângulo de Wigner (fórmula): {theta_deg:.1f}°")
```
