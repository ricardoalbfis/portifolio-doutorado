```python
import camb
import numpy as np
import matplotlib.pyplot as plt
print("CAMB:", camb.__version__)
pars = camb.CAMBparams()
pars.set_cosmology(
H0=67.4,
ombh2=0.0224,
omch2=0.120,
mnu=0.06,
omk=0,
tau=0.054
)
pars.InitPower.set_params(
As=2.1e-9,
ns=0.965
)
pars.set_for_lmax(
2500,
lens_potential_accuracy=1
)
results = camb.get_results(pars)
idade = results.physical_time(0)
print("Idade do Universo:", idade, "bilhões de anos")
powers = results.get_cmb_power_spectra(
pars,
CMB_unit='muK'
)
total = powers['total']
ell = np.arange(total.shape[0])
TT = total[:, 0]
EE = total[:, 1]
BB = total[:, 2]
TE = total[:, 3]
plt.figure(figsize=(9,6))
plt.plot(ell[2:], TT[2:])
plt.xlabel(r'$\ell$')
plt.ylabel(r'$D_\ell^{TT}\;[\mu K^2]$')
plt.xlim(2, 2500)
plt.show()
```
