# Exemplo fictício — Precessão do periélio (Schwarzschild)

Notebook de demonstração para testar a integração com o Colab. Calcula, de forma simplificada, a precessão do periélio prevista pela Relatividade Geral e compara com a órbita newtoniana.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parâmetros fictícios (unidades arbitrárias, G = c = 1)
M = 1.0        # massa central
a = 10.0       # semieixo maior
e = 0.6        # excentricidade
n_orbitas = 6

# Precessão relativística aproximada por órbita (fórmula clássica de Einstein)
delta_phi = 6 * np.pi * M / (a * (1 - e**2))
print(f"Precessão por órbita: {np.degrees(delta_phi):.4f} graus")
```

```python
theta = np.linspace(0, n_orbitas * 2 * np.pi, 4000)

def raio(theta, precessao_por_volta):
    phi = theta * (1 + precessao_por_volta / (2 * np.pi))
    return a * (1 - e**2) / (1 + e * np.cos(phi))

r_newton = raio(theta, 0)
r_gr = raio(theta, delta_phi)

x_n, y_n = r_newton * np.cos(theta), r_newton * np.sin(theta)
x_gr, y_gr = r_gr * np.cos(theta), r_gr * np.sin(theta)

plt.figure(figsize=(6, 6))
plt.plot(x_n, y_n, label="Órbita newtoniana", alpha=0.6)
plt.plot(x_gr, y_gr, label="Órbita com precessão (GR)", alpha=0.8)
plt.scatter([0], [0], color="black", s=40, label="Massa central")
plt.gca().set_aspect("equal")
plt.legend()
plt.title("Precessão do periélio — exemplo fictício")
plt.show()
```
